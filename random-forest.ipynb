```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons

# Generate dataset
X, y = make_moons(n_samples=1000, noise=0.2)

# Create DataFrame
df = {
    "X1": X[:, 0],
    "X2": X[:, 1],
    "y": y
}

dataset = pd.DataFrame(df)


print(dataset.head())
```

```python
import seaborn as sns

sns.scatterplot(x="X1", y="X2", data=dataset, hue="y")
plt.show()
```

```python
X_a = dataset.iloc[:,:-1]
Y_a = dataset["y"]
```

```python
from sklearn.model_selection import train_test_split

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    X_a,
    Y_a,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Model Create
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train Model
rf.fit(x_train, y_train)
```

```python

# Accuracy
print("Train Accuracy:", rf.score(x_train, y_train) * 100)
print("Test Accuracy:", rf.score(x_test, y_test) * 100)

#here problem is the model is doing overfitting so we have to handle them
```

```python

#We handle the overfiting of model through hyperparameters and check through cross validataion
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    min_samples_leaf=5,
    random_state=42
)

scores = cross_val_score(rf, x_train, y_train, cv=5)

print("CV Accuracy:", scores.mean() * 100)
```
